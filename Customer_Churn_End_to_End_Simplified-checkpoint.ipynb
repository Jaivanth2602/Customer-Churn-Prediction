## 4. Clean the data

Two problems to fix before anything else:

1. `TotalCharges` is stored as **text**, not a number (a few rows have blank spaces instead of a value) — I convert it to numeric and let blanks become `NaN` for now.
2. `Churn` is `"Yes"`/`"No"` text — I convert it to `1`/`0` so models can use it.
3. `customerID` is just an ID, not useful for prediction, so I drop it.

I'm *not* manually filling the missing `TotalCharges` values here — that happens later, inside the pipeline, so the fix only ever "sees" training data (avoids leaking test-set information).


````python
df = df.copy()  # work on a copy so re-running cells doesn't cause surprises

df["TotalCharges"] = pd.to_numeric(df["TotalCharges"], errors="coerce")   # text -> number, bad values -> NaN
df["Churn"] = df["Churn"].map({"Yes": 1, "No": 0})                        # Yes/No -> 1/0
df.drop(columns=["customerID"], inplace=True)                            # not predictive, just an ID

print("Shape after cleaning:", df.shape)
df.head()
````

````python
print("Missing values after conversion:")
display(df.isnull().sum().sort_values(ascending=False).head())
````

## 6. Statistical test — is Contract really linked to Churn?

The bar chart *looks* like contract type matters, but a chi-square test checks that statistically.

- **H0 (null hypothesis):** contract type and churn are independent (no real link)
- **H1:** they are associated

Chi-square is the right test here because both `Contract` and `Churn` are categories, not numbers.


````python
contract_table = pd.crosstab(df["Contract"], df["Churn"])
chi2, p_value, dof, expected = chi2_contingency(contract_table)

print("Chi-square statistic:", round(chi2, 3))
print("p-value:", p_value)

if p_value < 0.05:
    print("Result: reject H0 -> contract type and churn ARE associated.")
else:
    print("Result: fail to reject H0 -> not enough evidence of association.")
````

**How I'd explain this:** "A p-value under 0.05 means the pattern I saw in the chart is very unlikely to be random chance — so contract type genuinely relates to churn, it's not just noise in this sample."


## 7. Prepare the data for machine learning

I'll train two models and compare them:

1. **Logistic Regression** — simple, fast, and easy to explain (each feature gets a weight)
2. **Random Forest** — a bit of a "black box", but usually more accurate because it can capture non-linear patterns

**Why a pipeline?** So every preprocessing step (filling missing values, scaling numbers, one-hot encoding text columns) happens automatically and consistently on both train and test data — no manual, error-prone repeating of steps.


````python
X = df.drop(columns=["Churn"])
y = df["Churn"]

# Split columns into numeric vs text/categorical, since they need different treatment
numeric_features = X.select_dtypes(exclude="object").columns.tolist()
categorical_features = X.select_dtypes(include="object").columns.tolist()

print("Numeric features:", numeric_features)
print("\nCategorical features:", categorical_features)
````

````python
# Numeric columns: fill missing values with the median, then scale to mean 0 / std 1
numeric_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

# Categorical columns: fill missing values with the most common category, then one-hot encode
categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("onehot", OneHotEncoder(handle_unknown="ignore"))
])

# Apply the right pipeline to the right columns
preprocessor = ColumnTransformer([
    ("num", numeric_pipeline, numeric_features),
    ("cat", categorical_pipeline, categorical_features)
])

# 80% train, 20% test. stratify=y keeps the churn/no-churn ratio the same in both sets
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y
)

print("Train size:", X_train.shape[0], "| Test size:", X_test.shape[0])
````

**How I'd explain this:** "I split columns into numeric and categorical because they need different fixes — numeric columns get median-filled and scaled, text columns get one-hot encoded. I wrapped it all in a `ColumnTransformer` + `Pipeline` so the exact same steps run on train and test data automatically, and I stratified the split so both sets have the same churn rate."


## 8. Model 1 — Logistic Regression (baseline)


````python
logistic_model = Pipeline([
    ("preprocessor", preprocessor),
    ("model", LogisticRegression(max_iter=1000))   # max_iter raised so it fully converges
])

logistic_model.fit(X_train, y_train)

log_pred = logistic_model.predict(X_test)
log_prob = logistic_model.predict_proba(X_test)[:, 1]   # probability of churn (class 1)

print(classification_report(y_test, log_pred))
print("ROC-AUC:", round(roc_auc_score(y_test, log_prob), 3))
````

````python
ConfusionMatrixDisplay.from_predictions(y_test, log_pred)
plt.title("Logistic Regression — Confusion Matrix")
plt.show()
````

**How I'd explain this:** "Precision tells me: of everyone I flagged as a churner, how many really were. Recall tells me: of everyone who actually churned, how many I caught. ROC-AUC is an overall score of how well the model ranks churners above non-churners, regardless of the threshold."


## 9. Model 2 — Random Forest

A Random Forest is just many decision trees voting together. I use a few settings to keep it from overfitting:
- `max_depth=10` → stops trees from memorizing the training data
- `min_samples_leaf=3` → each leaf needs at least 3 examples
- `class_weight="balanced"` → since churners are the minority class, this makes the model pay more attention to them


````python
random_forest = Pipeline([
    ("preprocessor", preprocessor),
    ("model", RandomForestClassifier(
        n_estimators=300,        # number of trees
        max_depth=10,            # limits how complex each tree gets
        min_samples_leaf=3,      # avoids tiny, overfit leaf nodes
        class_weight="balanced", # gives more weight to the minority (churn) class
        random_state=42
    ))
])

random_forest.fit(X_train, y_train)

rf_pred = random_forest.predict(X_test)
rf_prob = random_forest.predict_proba(X_test)[:, 1]

print(classification_report(y_test, rf_pred))
print("ROC-AUC:", round(roc_auc_score(y_test, rf_prob), 3))
````

````python
ConfusionMatrixDisplay.from_predictions(y_test, rf_pred)
plt.title("Random Forest — Confusion Matrix")
plt.show()
````

## 10. Compare the two models side by side


````python
def get_metrics(y_true, pred, prob):
    return {
        "Accuracy": accuracy_score(y_true, pred),
        "Precision": precision_score(y_true, pred),
        "Recall": recall_score(y_true, pred),
        "F1": f1_score(y_true, pred),
        "ROC-AUC": roc_auc_score(y_true, prob)
    }

results = pd.DataFrame({
    "Logistic Regression": get_metrics(y_test, log_pred, log_prob),
    "Random Forest": get_metrics(y_test, rf_pred, rf_prob)
}).T

results.round(3)
````

````python
RocCurveDisplay.from_predictions(y_test, log_prob, name="Logistic Regression")
RocCurveDisplay.from_predictions(y_test, rf_prob, name="Random Forest")
plt.title("ROC Curve Comparison")
plt.show()
````

### Which model do I pick?

I'm using **Random Forest** going forward, because:
- it usually gives a better ROC-AUC / recall balance on this kind of data
- it lets me pull out **feature importance** afterwards (which columns actually drove the prediction)

**How I'd explain this:** "Logistic Regression is my interpretable baseline — I always start simple. Random Forest usually wins on recall/ROC-AUC because it captures interactions logistic regression can't, and as a bonus it gives me feature importances for the business story."


## 11. Choosing a decision threshold

The model doesn't just say "Yes/No" — it gives a **probability** of churn (e.g. 0.72 = 72% chance). By default, anything ≥ 0.50 is labeled "will churn", but that 0.50 cutoff is arbitrary.

Lowering the threshold catches more real churners (higher recall) but also flags more people who wouldn't actually have left (lower precision). I check a range of thresholds to see that trade-off.


````python
thresholds = np.arange(0.30, 0.71, 0.05)
threshold_results = []

for t in thresholds:
    pred_at_t = (rf_prob >= t).astype(int)
    threshold_results.append({
        "Threshold": t,
        "Precision": precision_score(y_test, pred_at_t),
        "Recall": recall_score(y_test, pred_at_t),
        "F1": f1_score(y_test, pred_at_t)
    })

threshold_results = pd.DataFrame(threshold_results)
threshold_results.round(3)
````

````python
plt.plot(threshold_results["Threshold"], threshold_results["Precision"], marker="o", label="Precision")
plt.plot(threshold_results["Threshold"], threshold_results["Recall"], marker="o", label="Recall")
plt.plot(threshold_results["Threshold"], threshold_results["F1"], marker="o", label="F1")
plt.xlabel("Classification Threshold")
plt.ylabel("Score")
plt.title("Precision / Recall / F1 vs Threshold")
plt.legend()
plt.show()
````

I'm using **0.40** as the threshold for this project — it leans toward catching more churners (higher recall), which usually matters more than precision in retention (missing a churner costs more than a wasted retention offer).

In a real company, the threshold would be chosen using the actual cost of a retention offer vs. the value of a retained customer.

**How I'd explain this:** "I picked 0.40 instead of the default 0.50 because in churn prevention, missing a real churner is usually more expensive than wasting a discount on someone who wouldn't have left anyway — so I biased the model toward recall."


````python
chosen_threshold = 0.40

test_results = X_test.copy()
test_results["ActualChurn"] = y_test.values
test_results["ChurnProbability"] = rf_prob
test_results["PredictedChurn"] = (rf_prob >= chosen_threshold).astype(int)

test_results.head()
````

## 12. Feature importance — what does the model rely on?

This tells us which columns the Random Forest found most useful for separating churners from non-churners. It's about **predictive usefulness, not proof of cause** — e.g. "short tenure" predicts churn, but tenure itself doesn't *cause* someone to leave.


````python
feature_names = random_forest.named_steps["preprocessor"].get_feature_names_out()
importance_values = random_forest.named_steps["model"].feature_importances_

importance_df = (
    pd.DataFrame({"Feature": feature_names, "Importance": importance_values})
    .sort_values("Importance", ascending=False)
)

importance_df.head(15)
````

````python
top_features = importance_df.head(15).sort_values("Importance")

plt.figure(figsize=(8, 6))
plt.barh(top_features["Feature"], top_features["Importance"])
plt.xlabel("Importance")
plt.title("Top 15 Features Driving Churn Predictions")
plt.show()
````

## 13. Customer segmentation with K-Means

Different question now: instead of "will this customer churn?", it's **"what kind of customer is this?"**

I'm using 3 simple numeric columns so the segments stay easy to explain:
- `tenure`
- `MonthlyCharges`
- `TotalCharges`


````python
segment_features = ["tenure", "MonthlyCharges", "TotalCharges"]
segment_data = df[segment_features].fillna(df[segment_features].median())

segment_scaler = StandardScaler()   # K-Means needs scaled data, or big-number columns dominate
segment_scaled = segment_scaler.fit_transform(segment_data)

kmeans = KMeans(n_clusters=4, random_state=42, n_init=10)
df["Segment"] = kmeans.fit_predict(segment_scaled)

df[segment_features + ["Segment"]].head()
````

````python
segment_summary = (
    df.groupby("Segment")
      .agg(
          Customers=("Segment", "size"),
          AvgTenure=("tenure", "mean"),
          AvgMonthlyCharges=("MonthlyCharges", "mean"),
          AvgTotalCharges=("TotalCharges", "mean"),
          ChurnRate=("Churn", "mean")
      )
      .sort_values("ChurnRate", ascending=False)
)

segment_summary.round(2)
````

````python
plt.figure(figsize=(9, 6))
sns.scatterplot(data=df, x="tenure", y="MonthlyCharges", hue="Segment", palette="tab10", s=60)
plt.title("Customer Segments")
plt.xlabel("Tenure")
plt.ylabel("Monthly Charges")
plt.show()
````

## 14. Score every customer

Now I run the trained Random Forest on the **entire** dataset (not just the test set) so every customer gets a churn probability. I also use `TotalCharges` as a rough stand-in for "customer value" — this is *not* real Customer Lifetime Value, just a simple proxy for a portfolio project.


````python
model_features = df.drop(columns=["Churn", "Segment"])

df["ChurnProbability"] = random_forest.predict_proba(model_features)[:, 1]
df["CustomerValue"] = df["TotalCharges"]

df[["ChurnProbability", "CustomerValue", "Segment"]].head()
````

## 15. Turn scores into a retention plan

I combine **churn risk** with **customer value** into one priority label:

| | High value | Low value |
|---|---|---|
| **High risk** | High Priority | Medium Priority |
| **Low risk** | Loyal / Protect | Low Priority |


````python
high_risk = df["ChurnProbability"] >= chosen_threshold
high_value = df["CustomerValue"] >= df["CustomerValue"].median()

df["RetentionPriority"] = np.select(
    [high_risk & high_value, high_risk & ~high_value, ~high_risk & high_value],
    ["High Priority", "Medium Priority", "Loyal / Protect"],
    default="Low Priority"
)

priority_summary = (
    df.groupby("RetentionPriority")
      .agg(
          Customers=("RetentionPriority", "size"),
          AvgChurnProbability=("ChurnProbability", "mean"),
          AvgCustomerValue=("CustomerValue", "mean")
      )
      .sort_values("AvgChurnProbability", ascending=False)
)

priority_summary.round(2)
````

**How I'd explain this:** "This turns two numbers — risk and value — into one business-friendly label. It's a simple way to hand a marketing team a ranked list instead of raw probabilities they'd have to interpret themselves."


## 16. The actual retention call-list

Top 20 customers to contact first: highest churn risk, and among those, highest value.


````python
high_priority_customers = (
    df[df["RetentionPriority"] == "High Priority"]
    .sort_values(["ChurnProbability", "CustomerValue"], ascending=[False, False])
)

high_priority_customers[[
    "ChurnProbability", "CustomerValue", "Segment",
    "Contract", "tenure", "MonthlyCharges", "RetentionPriority"
]].head(20)
````
